# Tech Challenge — Machine Learning para Classificação de Tumores de Mama

## Problema

Neste trabalho será desenvolvida uma solução de **Machine Learning** para classificação binária de tumores de mama.

O objetivo é construir e avaliar modelos capazes de classificar uma amostra como **benigna ou maligna**, utilizando características numéricas extraídas de imagens digitalizadas de massas mamárias.

O desenvolvimento seguirá um fluxo completo de Machine Learning, iniciando pela exploração e compreensão dos dados, passando pela preparação, treinamento e comparação de diferentes algoritmos, até a seleção e avaliação final do modelo.

---

## Dataset

O conjunto de dados utilizado é o **Wisconsin Diagnostic Breast Cancer (WDBC)**, composto por **569 amostras**.

Cada amostra possui um identificador, um diagnóstico e **30 características numéricas** relacionadas às propriedades observadas nas imagens digitalizadas das massas mamárias.

A variável alvo é `Diagnosis`, representada originalmente por:

- **B — Benign:** Benigno
- **M — Malignant:** Maligno

Durante a preparação dos dados, o diagnóstico será convertido para representação binária:

- **0 — Benigno**
- **1 — Maligno**

## 4. Associação entre as características e o diagnóstico

Como as características são numéricas, será criada uma representação binária do diagnóstico:

- `0` = Benigno;
- `1` = Maligno.

Em seguida será calculada a correlação de cada característica com a variável alvo. O objetivo é identificar quais medidas apresentam maior associação linear com o diagnóstico.

Valores de correlação mais próximos de `1` ou `-1` indicam associação linear mais forte, enquanto valores próximos de `0` indicam pouca associação linear.


```python
df_analise = df.copy()

df_analise["Diagnosis_Binary"] = (
    df_analise["Diagnosis"]
    .map({
        "B": 0,
        "M": 1
    })
)

features = [
    coluna
    for coluna in df.columns
    if coluna not in ["ID", "Diagnosis"]
]

correlacoes = (
    df_analise[features + ["Diagnosis_Binary"]]
    .corr()["Diagnosis_Binary"]
    .drop("Diagnosis_Binary")
    .sort_values(
        key=abs,
        ascending=False
    )
)

correlacoes
```

```python
correlacoes_plot = correlacoes.sort_values()

plt.figure(figsize=(10, 9))

plt.barh(
    correlacoes_plot.index,
    correlacoes_plot.values
)

plt.axvline(
    x=0,
    linewidth=1
)

plt.xlabel("Correlação com Diagnosis (B=0, M=1)")
plt.ylabel("Característica")
plt.title("Correlação das Características com o Diagnóstico")

plt.tight_layout()
plt.show()
```

## 5. Preparação das variáveis para Machine Learning

Após a análise exploratória, os dados serão preparados para os algoritmos de classificação.

A coluna `ID` será removida por representar apenas um identificador, sem significado preditivo. A coluna `Diagnosis` será utilizada como variável alvo e convertida para formato binário:

- `0` = Benigno (`B`);
- `1` = Maligno (`M`).

Ao final desta etapa:

- `X` conterá as 30 características utilizadas pelos modelos;

```python
X = df.drop(columns=["ID", "Diagnosis"])

y = df["Diagnosis"].map({
    "B": 0,
    "M": 1
})

print("Dimensão de X:", X.shape)
print("Dimensão de y:", y.shape)

print("\nDistribuição do alvo:")
print(y.value_counts())

print("\nDistribuição percentual:")
print(
    y.value_counts(normalize=True)
     .mul(100)
     .round(2)
)
```

## 6. Separação em treinamento e teste

O dataset será dividido em dois conjuntos:

- **80% para treinamento (`X_train` e `y_train`)**;
- **20% para teste (`X_test` e `y_test`)**.

O parâmetro `stratify=y` preserva aproximadamente a proporção de casos benignos e malignos nos dois conjuntos.

O conjunto de teste permanecerá separado durante o desenvolvimento, comparação e otimização dos modelos. Ele será utilizado somente na avaliação final do modelo selecionado.

O `random_state=42` garante reprodutibilidade da divisão.


```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

print("\nDistribuição no treino:")
print(y_train.value_counts())
print(
    y_train.value_counts(normalize=True)
           .mul(100)
           .round(2)
)

print("\nDistribuição no teste:")
print(y_test.value_counts())
print(
    y_test.value_counts(normalize=True)
          .mul(100)
          .round(2)
)
```

## 7. Modelo 1 — Regressão Logística

A Regressão Logística será utilizada como primeiro modelo de classificação.

Como as características possuem escalas diferentes, será utilizado um `StandardScaler` antes do modelo. A padronização transforma as variáveis para escalas comparáveis e é realizada dentro de um `Pipeline`, evitando vazamento de informação entre treinamento e validação.

O pipeline executará:

1. padronização das características;
2. treinamento da Regressão Logística;
3. geração das classificações.


```python
 from sklearn.pipeline import Pipeline
 from sklearn.preprocessing import StandardScaler
 from sklearn.linear_model import LogisticRegression

 pipeline_lr = Pipeline([
     ("scaler", StandardScaler()),
     (
         "modelo",
         LogisticRegression(
             max_iter=1000,
             random_state=42
         )
     )
 ])
```

### 7.1 Validação cruzada estratificada

Para avaliar o modelo sem utilizar o conjunto de teste, será aplicada **validação cruzada estratificada com 5 folds**.

Os 455 registros de treinamento são divididos em cinco partes. Em cada rodada, quatro partes são utilizadas para treinamento e uma para validação. O processo é repetido até que todos os folds tenham sido utilizados como validação.

Serão avaliadas:

- **Accuracy** — proporção total de classificações corretas;
- **Precision** — entre os casos previstos como malignos, quantos realmente são malignos;
- **Recall** — entre os casos realmente malignos, quantos foram identificados;
- **F1-score** — equilíbrio entre Precision e Recall;
- **ROC AUC** — capacidade do modelo de discriminar as duas classes pelos scores produzidos.

Como a classe `1` representa malignidade, o **Recall** merece atenção especial.


```python
from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

metricas = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc"
}

resultado_lr = cross_validate(
    pipeline_lr,
    X_train,
    y_train,
    cv=cv,
    scoring=metricas,
    n_jobs=-1
)
```

### 7.2 Função para apresentação das métricas

Para facilitar a comparação entre os modelos, será criada uma função que apresenta os resultados obtidos nos cinco folds.

Além dos valores individuais, serão mostrados a média e o desvio padrão de cada métrica. A média resume o desempenho e o desvio ajuda a avaliar a estabilidade entre as diferentes divisões dos dados.


```python
import numpy as np

def mostrar_resultados(nome, resultado):
    print(nome)
    print("-" * len(nome))

    for metrica in [
        "accuracy",
        "precision",
        "recall",
        "f1",
        "roc_auc"
    ]:
        valores = resultado[f"test_{metrica}"]

        print(
            f"{metrica:10s}: "
            f"{np.round(valores, 4)} "
            f"| média = {valores.mean():.4f} "
            f"| desvio = {valores.std():.4f}"
        )

mostrar_resultados(
    "Regressão Logística",
    resultado_lr
)
```

### Resultado da Regressão Logística

A Regressão Logística foi avaliada utilizando validação cruzada
estratificada com 5 folds sobre o conjunto de treinamento.

Os resultados médios obtidos foram:

- Accuracy: **0,9736**
- Precision: **0,9771**
- Recall: **0,9529**
- F1-score: **0,9640**
- ROC AUC: **0,9958**

Os resultados indicam elevada capacidade de classificação e
discriminação entre tumores benignos e malignos.

O Recall de aproximadamente 95,3% indica que o modelo identificou a
grande maioria dos casos malignos durante a validação cruzada. Essa
métrica é particularmente relevante neste problema, pois falsos
negativos correspondem a casos malignos classificados como benignos.

O ROC AUC médio de aproximadamente 0,996, acompanhado de baixo desvio
entre os folds, indica elevada capacidade discriminativa e boa
estabilidade do modelo nas diferentes divisões dos dados.

Esses resultados são consistentes com a análise exploratória, na qual
diversas características apresentaram forte associação com a variável
alvo.

## 8. Modelo 2 — Árvore de Decisão

A Árvore de Decisão classifica as amostras por meio de regras hierárquicas construídas a partir das características.

Diferentemente da Regressão Logística e do KNN, árvores não dependem da escala das variáveis; portanto, não é necessário aplicar `StandardScaler`.

Inicialmente será utilizado o modelo com seus parâmetros básicos para estabelecer um **baseline**, usando a mesma validação cruzada e as mesmas métricas.

O KNN classifica uma nova amostra observando as classes das amostras mais próximas.

Como o algoritmo utiliza distâncias, características com escalas maiores poderiam dominar o cálculo. Por isso, o `StandardScaler` é fundamental e será incluído no `Pipeline`.

## Seleção do Limiar de Decisão (Threshold)

Após a seleção da Regressão Logística como modelo final, será analisado o limiar de decisão utilizado para converter as probabilidades produzidas pelo modelo em classes.

Por padrão, a Regressão Logística utiliza um threshold de **0,50**:

- probabilidade de malignidade menor que 0,50 → **Benigno**;
- probabilidade de malignidade maior ou igual a 0,50 → **Maligno**.

Como um falso negativo representa um caso maligno classificado como benigno, será analisada a possibilidade de utilizar um threshold diferente de 0,50.

Para evitar vazamento de informação, essa escolha será realizada utilizando **somente o conjunto de treinamento**.

Serão utilizadas previsões *out-of-fold* obtidas por validação cruzada. Dessa forma, a probabilidade atribuída a cada amostra de treinamento será produzida por um modelo que não utilizou aquela amostra durante seu próprio treinamento.

O conjunto de teste permanecerá fora desta etapa e será utilizado posteriormente para a avaliação final.

```python
from sklearn.model_selection import cross_val_predict

y_prob_oof = cross_val_predict(
    pipeline_lr,
    X_train,
    y_train,
    cv=cv,
    method="predict_proba"
)[:, 1]

print("Quantidade de probabilidades geradas:", len(y_prob_oof))
```

```python
import numpy as np
import pandas as pd

from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

thresholds = np.arange(0.10, 0.91, 0.05)

resultados_threshold = []

for threshold in thresholds:

    y_pred_threshold = (
        y_prob_oof >= threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_train,
        y_pred_threshold
    ).ravel()

    resultados_threshold.append({
        "Threshold": threshold,
        "Accuracy": accuracy_score(
            y_train,
            y_pred_threshold
        ),
        "Precision": precision_score(
            y_train,
            y_pred_threshold
        ),
        "Recall": recall_score(
            y_train,
            y_pred_threshold
        ),
        "F1": f1_score(
            y_train,
            y_pred_threshold
        ),
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    })

df_thresholds_oof = pd.DataFrame(
    resultados_threshold
)

df_thresholds_oof.round(4)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

plt.plot(
    df_thresholds_oof["Threshold"],
    df_thresholds_oof["Precision"],
    marker="o",
    label="Precision"
)

plt.plot(
    df_thresholds_oof["Threshold"],
    df_thresholds_oof["Recall"],
    marker="o",
    label="Recall"
)

plt.plot(
    df_thresholds_oof["Threshold"],
    df_thresholds_oof["F1"],
    marker="o",
    label="F1-score"
)

plt.xlabel("Threshold")
plt.ylabel("Métrica")
plt.title(
    "Impacto do Threshold - Validação Out-of-Fold"
)

plt.ylim(0.80, 1.02)

plt.grid(alpha=0.3)
plt.legend()

plt.show()
```

```python
candidatos = df_thresholds_oof[
    df_thresholds_oof["Recall"] >= 0.95
]

candidatos.round(4)
```

```python
threshold_final = 0.30

print(f"Threshold final selecionado: {threshold_final:.2f}")
```

### Escolha do Threshold

A análise das previsões *out-of-fold* permitiu avaliar diferentes limiares de decisão utilizando somente o conjunto de treinamento.

Como um **falso negativo** representa um caso maligno classificado como benigno, foi definido como critério manter um **Recall mínimo de 95% para a classe maligna**.

Entre os thresholds que atenderam a esse requisito, foi analisado o equilíbrio entre **Recall, Precision e F1-score**.

#### Threshold padrão de 0,50

- Recall: **95,29%**
- Precision: **97,59%**
- F1-score: **96,43%**
- Falsos negativos: **8**
- Falsos positivos: **4**

#### Threshold de 0,30

- Recall: **97,65%**
- Precision: **95,40%**
- F1-score: **96,51%**
- Falsos negativos: **4**
- Falsos positivos: **8**

Ao reduzir o threshold de **0,50 para 0,30**, o Recall aumentou de **95,29% para 97,65%** e a quantidade de falsos negativos caiu de **8 para 4**.

Em contrapartida, os falsos positivos aumentaram de **4 para 8**. Esse comportamento representa um *trade-off* entre Recall e Precision.

Neste experimento, optou-se por priorizar a identificação dos casos malignos. Por esse motivo, foi selecionado o **threshold de 0,30**.

A escolha foi realizada utilizando exclusivamente as previsões *out-of-fold* do conjunto de treinamento, sem utilizar o conjunto de teste.

> **Observação:** o threshold de 0,30 representa uma decisão metodológica deste experimento e não deve ser interpretado como um limiar clinicamente validado.

Em seguida, essas probabilidades serão convertidas em classes utilizando o **threshold de 0,30**, previamente selecionado por meio da validação *out-of-fold* realizada exclusivamente no conjunto de treinamento:

- probabilidade de malignidade menor que **0,30** → **Benigno (0)**;

A curva ROC apresentou AUC de **0,996**, indicando excelente capacidade de discriminação entre as classes neste experimento.

Apesar dos resultados elevados, este estudo possui finalidade acadêmica. O desempenho observado no dataset WDBC não representa validação clínica do modelo. Uma aplicação real exigiria validação externa independente, análise em diferentes populações e avaliação clínica adequada antes de qualquer utilização em processos de diagnóstico ou apoio à decisão médica.
