### 1.1 As variaveis gender, department e depression sao categoricas, devemos transforma-las

```python
df[["Gender","Department","Depression"]].head()
```

#### 1.1.1 As variaveis gender e department sao dummies, transformando

```python
df=pd.get_dummies(df,columns=["Gender","Department"],drop_first=True)
```

```python
df.head()
```

#### 1.1.2 Agora selecionamos as variaveis booleanas e colocamos como inteiras

```python
bool_cols = df.select_dtypes(include='bool').columns
df[bool_cols] = df[bool_cols].astype(int)
```

```python
df.head()
```

### 1.2 Para analise e modelagem, o id do estudante n eh uma variavel interessante, podemos dropar
